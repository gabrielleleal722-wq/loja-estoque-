```python
# Sistema simples de estoque de loja
# Demonstra os principais conceitos de Python usando POO

from abc import ABC, abstractmethod


# Classe abstrata (abstração)
class Produto(ABC):
    def __init__(self, nome, preco, quantidade):
        # atributos "privados" (encapsulamento)
        self.__nome = nome
        self.__preco = float(preco)
        self.__quantidade = int(quantidade)

    def get_nome(self):
        return self.__nome

    def get_preco(self):
        return self.__preco

    def get_quantidade(self):
        return self.__quantidade

    def adicionar_estoque(self, quantidade):
        self.__quantidade += quantidade

    def remover_estoque(self, quantidade):
        if quantidade > self.__quantidade:
            print(f"Estoque insuficiente! Faltam {abs(self.__quantidade - quantidade)} unidade(s).")
        else:
            self.__quantidade -= quantidade

    def valor_total(self):
        return self.__preco * self.__quantidade

    # método abstrato -> a subclasse é obrigada a implementar
    @abstractmethod
    def aplicar_desconto(self, percentual):
        pass

    def __str__(self):
        return f"{self.__nome} | R${self.__preco:.2f} | Qtd: {self.__quantidade}"


# Herança: ProdutoEletronico é um tipo de Produto
class ProdutoEletronico(Produto):
    def __init__(self, nome, preco, quantidade, garantia_meses):
        super().__init__(nome, preco, quantidade)
        self.garantia_meses = garantia_meses

    def aplicar_desconto(self, percentual):
        novo_preco = self.get_preco() - (self.get_preco() * percentual / 100)
        self._Produto__preco = novo_preco
        print(f"Novo preço de {self.get_nome()}: R${novo_preco:.2f}")

    def __str__(self):
        return f"{super().__str__()} | Garantia: {self.garantia_meses} meses"


# Lista que guarda os produtos cadastrados
estoque = []

# Cadastro inicial de exemplo
estoque.append(ProdutoEletronico("Fone de Ouvido", 150.0, 10, 12))
estoque.append(ProdutoEletronico("Mouse", 45.0, 20, 6))


def listar_produtos():
    if len(estoque) == 0:
        print("Estoque vazio.")
        return

    for i in range(len(estoque)):
        print(f"{i} - {estoque[i]}")


def buscar_produto(nome):
    for produto in estoque:
        if nome.lower() in produto.get_nome().lower():
            return produto
    return None


# Loop principal do programa
opcao = -1

while opcao != 0:
    print("\n--- MENU ---")
    print("1 - Cadastrar produto")
    print("2 - Listar produtos")
    print("3 - Adicionar estoque")
    print("4 - Remover estoque")
    print("5 - Remover produto")
    print("6 - Buscar produto")
    print("7 - Aplicar desconto")
    print("8 - Valor total do estoque")
    print("0 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        nome = input("Nome: ")
        preco = float(input("Preço: "))
        quantidade = int(input("Quantidade: "))
        garantia = int(input("Garantia (meses): "))
        estoque.append(ProdutoEletronico(nome, preco, quantidade, garantia))
        print("Produto cadastrado!")

    elif opcao == 2:
        listar_produtos()

    elif opcao == 3:
        listar_produtos()
        indice = int(input("Índice do produto: "))
        if indice >= 0 and indice < len(estoque):
            qtd = int(input("Quantidade a adicionar: "))
            estoque[indice].adicionar_estoque(qtd)

    elif opcao == 4:
        listar_produtos()
        indice = int(input("Índice do produto: "))
        if indice >= 0 and indice < len(estoque):
            qtd = int(input("Quantidade a remover: "))
            estoque[indice].remover_estoque(qtd)

    elif opcao == 5:
        listar_produtos()
        indice = int(input("Índice do produto a remover: "))
        if indice >= 0 and indice < len(estoque):
            removido = estoque.pop(indice)
            print(f"'{removido.get_nome()}' removido do estoque.")

    elif opcao == 6:
        nome = input("Nome do produto: ")
        produto = buscar_produto(nome)
        if produto is not None:
            print(f"Encontrado: {produto}")
        else:
            print("Produto não encontrado.")

    elif opcao == 7:
        listar_produtos()
        indice = int(input("Índice do produto: "))
        if indice >= 0 and indice < len(estoque):
            percentual = float(input("Percentual de desconto: "))
            estoque[indice].aplicar_desconto(percentual)

    elif opcao == 8:
        total = 0
        for produto in estoque:
            total += produto.valor_total()
        print(f"Valor total do estoque: R${total:.2f}")

    elif opcao == 0:
        print("Encerrando o sistema...")

    else:
        print("Opção inválida!")
```
