```python
"""
Funções auxiliares que trabalham sobre a lista de estoque.
Não pertencem diretamente a nenhuma classe do domínio.
"""


def listar_produtos(estoque):
    """Mostra todos os produtos cadastrados, numerados por índice."""
    if len(estoque) == 0:
        print("Estoque vazio.")
        return

    for i in range(len(estoque)):
        print(f"{i} - {estoque[i]}")


def buscar_produto(estoque, nome):
    """Procura um produto pelo nome (busca parcial, sem diferenciar maiúsculas)."""
    for produto in estoque:
        if nome.lower() in produto.get_nome().lower():
            return produto
    return None


def valor_total_estoque(estoque):
    """Soma o valor total (preço x quantidade) de todos os produtos."""
    total = 0
    for produto in estoque:
        total += produto.valor_total()
    return total
```
